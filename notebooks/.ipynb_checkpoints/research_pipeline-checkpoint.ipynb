This notebook walks the code that is already implemented and used by the experiment runners. Each stage imports the repository and calls those functions. It does not reimplement feature math, splits, training, or backtests.

Smoke results are plumbing checks. They are not evidence of tradable alpha. That matches the current status of the v2 path: the stages run and write metrics, and the economic result on a short tail is not a promotion decision.

2. Feature frame from `DataPreprocessor`
3. Triple-barrier labels and sample weights on a short window
4. Multi-horizon dataset (stage 0)
5. Purged walk-forward splits, plus the older `PurgedTimeSeriesSplit`
6. Base LSTM training and out-of-fold predictions
7. Meta-labels, meta-labeler, and directional alpha backtest on validation only
8. Sharpe statistics and zero/momentum forecast baselines

## Raw data on disk

The collector is not called. These cells only describe the Lighter CSV exports the loaders already read, plus the side-data files that are stored and not joined into model features.


````python
def _csv_inventory(path: Path) -> dict:
    line_count = 0
    first = last = ""
    with path.open("r", encoding="utf-8") as handle:
        for line in handle:
            line_count += 1
            if line_count == 1:
                first = line.strip()
            last = line.strip()
    return {
        "path": str(path.resolve().relative_to(ROOT)),
        "bytes": path.stat().st_size,
        "lines": line_count,
        "first": first[:120],
        "last": last[:120],
    }

raw_rows = [_csv_inventory(path) for path in sorted(Path("data/raw").glob("ETHUSDT-*-lighter-*.csv"))]
display(pd.DataFrame(raw_rows))

side_rows = []
for path in sorted(Path("data/lighter").iterdir()):
    if not path.is_file():
        continue
    row = {"path": str(path.resolve().relative_to(ROOT)), "bytes": path.stat().st_size, "suffix": path.suffix}
    if path.suffix == ".csv":
        header = path.open("r", encoding="utf-8").readline().strip()
        row["header"] = header[:160]
        row["lines"] = sum(1 for _ in path.open("r", encoding="utf-8"))
    side_rows.append(row)
display(pd.DataFrame(side_rows))
````

## Feature generation

`DataPreprocessor.get_base_dataset` is the feature frame used by both dataset builders. It reads every matching Lighter OHLCV file, builds time bars, and adds the configured families: OHLCV, returns, realized volatility, volume, fractional difference, entropy, CUSUM, the simplified structural-break flag, and the volatility regime.

The fractional-difference call below is on a short tail so the fixed-width routine and its diagnostics are visible. The manifest diagnostics come from the full frame the preprocessor just built.

Activity bars use `construct_activity_bars` on a short raw slice. Thresholds are the median volume and median dollar volume of that slice, so the demo produces more than one bar. The model dataset stays on time bars.


````python
import time

from src.data.features_all import DataPreprocessor, bar_coverage_report, construct_activity_bars

preprocessor = DataPreprocessor(
    data_dir=base_config["data"]["dir"],
    granularities=[base_config["data"]["granularity"]],
    feature_config=base_config.get("features") or {},
    bar_config=base_config.get("bars") or {},
)
started = time.perf_counter()
features_df, targets_df = preprocessor.get_base_dataset(granularity=base_config["data"]["granularity"])
print(f"feature rows={len(features_df)} columns={len(features_df.columns)} seconds={time.perf_counter() - started:.1f}")
print("feature_columns", preprocessor.get_feature_cols())
print("feature_windows", preprocessor.feature_window_bars(base_config["data"]["granularity"]))
manifest = preprocessor.feature_manifest()
print("families", manifest["families"])
print("fracdiff_manifest", manifest["fracdiff"])
print("bar_manifest", preprocessor.bar_manifest())
display(features_df.tail(5))

tail = features_df.tail(300)
plot_columns = ["close", "log_return", "return_vol_24h", "fracdiff_close", "return_entropy_24h"]
fig, axes = plt.subplots(len(plot_columns), 1, figsize=(10, 9), sharex=True)
for axis, column in zip(axes, plot_columns):
    axis.plot(tail.index, tail[column], linewidth=0.8)
    axis.set_ylabel(column)
axes[0].set_title("Last 300 feature rows")
fig.tight_layout()
plt.show()

close_tail = targets_df["close"].tail(400)
diffed = preprocessor.fracdiff_fixed_width(
    close_tail,
    d=float(base_config["features"]["frac_diff_order"]),
    thres=float(base_config["features"]["frac_diff_threshold"]),
)
print("tail_fracdiff", preprocessor.fracdiff_diagnostics(
    close_tail,
    diffed,
    mode="fixed_width",
    order=float(base_config["features"]["frac_diff_order"]),
    threshold=float(base_config["features"]["frac_diff_threshold"]),
))

price_frame = preprocessor.load_price_data()[base_config["data"]["granularity"]]
price_slice = price_frame.tail(800).reset_index(drop=True)
volume_threshold = max(float(price_slice["volume"].quantile(0.5)), 1e-6)
dollar_threshold = max(float((price_slice["close"] * price_slice["volume"]).quantile(0.5)), 1e-6)
volume_bars = construct_activity_bars(price_slice, bar_type="volume", threshold=volume_threshold)
dollar_bars = construct_activity_bars(price_slice, bar_type="dollar", threshold=dollar_threshold)
print("volume_bars", bar_coverage_report(volume_bars, granularity="5m", bar_type="volume", source_rows=len(price_slice)))
print("dollar_bars", bar_coverage_report(dollar_bars, granularity="5m", bar_type="dollar", source_rows=len(price_slice)))
display(volume_bars[["timestamp", "open", "high", "low", "close", "volume"]].head(5))
display(dollar_bars[["timestamp", "open", "high", "low", "close", "volume"]].head(5))
````

## Labels and sample weights

`create_labels` and `sample_weights_from_labels` are the triple-barrier routines used by the older Lighter dataset builder. They are Python loops, so this cell uses 400 bars and a 24-bar timeout. The v2 meta-labels are built later from the multi-horizon dataset, not from this window.

`build_label_spans` and `average_uniqueness` are the span tools the v2 sample-weight path uses.


````python
from src.features.labeling import create_labels, sample_weights_from_labels
from src.features.sample_weights import average_uniqueness, build_label_spans

label_window = targets_df.tail(400).reset_index(drop=True)
labeled = create_labels(
    label_window,
    price_col="close",
    kappa=2.0,
    timeout=24,
    method="adaptive",
    volatility_window=24,
)
weights = sample_weights_from_labels(
    labeled["y_dir"],
    labeled["hit_times"],
    labeled["returns"],
    volatility_window=24,
)
span_end = np.arange(len(labeled)) + labeled["hit_times"].fillna(24).astype(int).clip(lower=0).to_numpy()
spans = build_label_spans(np.arange(len(labeled)), span_end)
uniqueness = average_uniqueness(spans["label_span_start_idx"], spans["label_span_end_idx"])
print("y_dir", labeled["y_dir"].value_counts(dropna=False).to_dict())
print("hit_times", labeled["hit_times"].describe().to_dict())
print("weight_sum", float(weights.sum()), "weight_head", weights.head(3).tolist())
print("average_uniqueness_mean", float(np.mean(uniqueness)), "effective_sample_size", float(np.sum(uniqueness)))
display(labeled[["close", "y_dir", "y_ret", "hit_times"]].head(8))
````

## Cross-validation

`purged_walk_forward_splits` is the splitter the v2 trial uses. Training for each fold stops before the validation block, with the configured purge. A gap and a held-out test block sit after development. This notebook records that test range and does not score it unless `SCORE_HELD_OUT_TEST` is turned on.

`PurgedTimeSeriesSplit` lives on the older trainer and is what `EnsemblePredictor` references. It is a different splitter. The cell calls it so the index pattern is visible. It is not the fold generator for the training cell below.


````python
from src.experiments.meta_labeling_mvp import purged_walk_forward_splits
from src.training.trainer import PurgedTimeSeriesSplit

splits = purged_walk_forward_splits(int(dataset["X"].shape[0]), active_config["validation"])
fold_rows = []
for fold in splits["folds"]:
    fold_rows.append({
        "fold": int(fold["fold"]),
        "train_start": int(fold["train"][0]),
        "train_end": int(fold["train"][-1]),
        "train_count": int(len(fold["train"])),
        "validation_start": int(fold["validation"][0]),
        "validation_end": int(fold["validation"][-1]),
        "validation_count": int(len(fold["validation"])),
    })
display(pd.DataFrame(fold_rows))
print({
    "purge_bars": int(splits["purge_bars"]),
    "embargo_bars": int(splits["embargo_bars"]),
    "development": int(len(splits["development"])),
    "gap": int(len(splits["gap"])),
    "test": int(len(splits["test"])),
    "test_start": int(splits["test"][0]) if len(splits["test"]) else None,
    "test_end": int(splits["test"][-1]) if len(splits["test"]) else None,
})

def _block(indices):
    if len(indices) == 0:
        return None
    start = int(indices[0])
    return (start, int(indices[-1]) - start + 1)

fig, axis = plt.subplots(figsize=(10, 3))
colors = {"train": "#4C78A8", "validation": "#F58518", "gap": "#E45756", "test": "#54A24B"}
labels_used = []
for row, fold in enumerate(splits["folds"]):
    for name in ("train", "validation"):
        block = _block(fold[name])
        if block is None:
            continue
        axis.broken_barh([block], (row, 0.8), facecolors=colors[name], label=name if name not in labels_used else None)
        labels_used.append(name)
for offset, name in enumerate(("gap", "test")):
    block = _block(splits[name])
    if block is None:
        continue
    axis.broken_barh([block], (len(splits["folds"]) + offset, 0.8), facecolors=colors[name], label=name)
axis.set_yticks(list(range(len(splits["folds"]) + 2)))
axis.set_yticklabels([f"fold {fold['fold']}" for fold in splits["folds"]] + ["gap", "test"])
axis.set_xlabel("sample index")
axis.set_title("Purged walk-forward coverage")
axis.legend(loc="upper right")
fig.tight_layout()
plt.show()

older_cv = PurgedTimeSeriesSplit(
    n_splits=2,
    purge_bars=int(active_config["validation"]["purge_bars"]),
    embargo_bars=int(active_config["validation"]["embargo_bars"]),
)
older_splits = older_cv.split(dataset["X"], dataset["y_ret"])
print("PurgedTimeSeriesSplit is the older trainer splitter, not the v2 fold generator.")
for index, (train_idx, val_idx) in enumerate(older_splits):
    print({
        "fold": index,
        "train": int(len(train_idx)),
        "validation": int(len(val_idx)),
        "train_start": int(train_idx[0]),
        "train_end": int(train_idx[-1]),
        "validation_start": int(val_idx[0]),
        "validation_end": int(val_idx[-1]),
    })
````

## Base model training

For each walk-forward fold the notebook calls `train_base_model` and `predict_base_model`, then trains one more base model on the development indices. That is the same order as a v2 trial. Out-of-fold rows are the only rows passed into meta-labeling.


````python
from src.experiments.meta_labeling_mvp import _prediction_metrics, predict_base_model, train_base_model

oof_frames = []
fold_histories = []
for fold in splits["folds"]:
    model, history = train_base_model(dataset, fold["train"], active_config, device)
    prediction = predict_base_model(model, dataset, fold["validation"], device)
    prediction["fold"] = int(fold["fold"])
    oof_frames.append(prediction)
    fold_histories.append({
        "fold": int(fold["fold"]),
        "final_loss": float(history["final_loss"]),
        "training_losses": [float(value) for value in history["training_losses"]],
        "metrics": _prediction_metrics(prediction, dataset["horizon_names"]),
    })
    print("fold", fold["fold"], "final_loss", history["final_loss"])

final_model, final_history = train_base_model(dataset, splits["development"], active_config, device)
print("development_final_loss", final_history["final_loss"])
assert all(math.isfinite(item["final_loss"]) for item in fold_histories)
assert math.isfinite(float(final_history["final_loss"]))

oof_predictions = pd.concat(oof_frames, ignore_index=True).sort_values("timestamp")
preview_columns = ["timestamp", "fold", "close"]
for horizon in dataset["horizon_names"]:
    preview_columns.extend([
        f"{horizon}_pred_return",
        f"{horizon}_true_return",
        f"{horizon}_direction_prob",
        f"{horizon}_true_direction",
    ])
display(oof_predictions[preview_columns].head(8))
print("oof_rows", len(oof_predictions))
print("fold_metrics", json.dumps(fold_histories, default=str)[:4000])
````

## Meta-labeling

Candidate signals, triple-barrier meta-labels, the meta-labeler, and the probability tables are the functions a v2 trial runs on its out-of-fold predictions.


````python
from src.experiments.meta_labeling_mvp import (
    add_meta_labels,
    add_meta_probabilities,
    candidate_signals,
    fit_meta_labeler,
    meta_threshold_table,
    predicted_edge_bucket_table,
    probability_bucket_table,
)

candidates = add_meta_labels(candidate_signals(oof_predictions, dataset, active_config), dataset, active_config)
meta_model = fit_meta_labeler(candidates, active_config, dataset["horizon_names"], device)
candidates = add_meta_probabilities(candidates, meta_model)

print("meta_model", meta_model.manifest())
print("candidates", len(candidates), "is_candidate", int(candidates["is_candidate"].sum()))
print("meta_label_counts", candidates["meta_label"].value_counts(dropna=False).to_dict())
print("sample_weight_diagnostics", candidates.attrs.get("sample_weight_diagnostics"))
display(pd.DataFrame(probability_bucket_table(candidates)))
display(pd.DataFrame(predicted_edge_bucket_table(candidates)))
display(pd.DataFrame(meta_threshold_table(candidates)))
label_columns = [
    "timestamp", "horizon", "side", "is_candidate", "expected_edge_bps",
    "meta_label", "meta_prob", "label_net_return",
]
display(candidates[[column for column in label_columns if column in candidates.columns]].head(8))
````

## Directional alpha backtest

`run_alpha_backtest` turns meta-probabilities into a validation trade list using the thresholds in `alpha_backtest`. The plot is cumulative net PnL of those validation trades. An empty trade list is a real outcome when the smoke model never clears the edge and probability gates.


````python
from src.experiments.meta_labeling_mvp import alpha_diagnostics, run_alpha_backtest

alpha_metrics, trades = run_alpha_backtest(candidates, active_config)
diagnostics = alpha_diagnostics(candidates, active_config)
headline = {
    key: alpha_metrics.get(key)
    for key in [
        "trades", "net_pnl", "gross_pnl", "fees", "hit_ratio", "win_rate",
        "max_drawdown", "turnover", "coverage", "horizon_distribution", "side_distribution",
    ]
}
print("alpha_backtest", headline)
print("diagnostics_keys", sorted(diagnostics))
if trades.empty:
    print("No validation trades at the configured thresholds.")
else:
    display(trades.head(8))
    fig, axis = plt.subplots(figsize=(10, 3))
    axis.plot(pd.to_datetime(trades["timestamp"]), trades["cumulative_net_pnl"])
    axis.set_title("Validation cumulative net PnL")
    axis.set_ylabel("net PnL")
    fig.autofmt_xdate()
    fig.tight_layout()
    plt.show()
````

## Held-out test split

The runner scores the test block only after a trial is selected on validation. The default leaves that block untouched. Setting `SCORE_HELD_OUT_TEST` runs the same public prediction, meta-label, and backtest functions on `splits["test"]`.


````python
print({
    "test_count": int(len(splits["test"])),
    "test_start": int(splits["test"][0]) if len(splits["test"]) else None,
    "test_end": int(splits["test"][-1]) if len(splits["test"]) else None,
    "gap_count": int(len(splits["gap"])),
    "scored": bool(SCORE_HELD_OUT_TEST),
})
if SCORE_HELD_OUT_TEST and len(splits["test"]):
    test_predictions = predict_base_model(final_model, dataset, splits["test"], device)
    test_candidates = add_meta_labels(candidate_signals(test_predictions, dataset, active_config), dataset, active_config)
    test_candidates = add_meta_probabilities(test_candidates, meta_model)
    test_metrics, test_trades = run_alpha_backtest(test_candidates, active_config)
    print("test_trades", test_metrics.get("trades"), "test_net_pnl", test_metrics.get("net_pnl"))
    display(test_trades.head(8))
````

## Backtest statistics

Sharpe and the probabilistic Sharpe use the validation trade returns when at least two finite returns exist. Deflated Sharpe needs two or more recorded trial Sharpes; one smoke trial returns the unavailable payload from `deflated_sharpe_ratio`.

The CSCV call uses a synthetic 32-by-3 return matrix. It exercises `probability_of_backtest_overfitting`. Those columns are not model trials.


````python
from src.evaluation.backtest_stats import (
    deflated_sharpe_ratio,
    probabilistic_sharpe_ratio,
    probability_of_backtest_overfitting,
    sharpe_ratio,
)

if trades.empty:
    trade_returns = np.asarray([], dtype=float)
else:
    trade_returns = trades["net_return"].to_numpy(dtype=float)
sharpe_payload = sharpe_ratio(trade_returns, granularity="5m")
psr_payload = probabilistic_sharpe_ratio(trade_returns)
trial_sharpe = sharpe_payload["per_period"]["value"] if sharpe_payload["per_period"]["available"] else 0.0
dsr_payload = deflated_sharpe_ratio(trade_returns, [trial_sharpe], trial_count=1)
print("sharpe", sharpe_payload)
print("probabilistic_sharpe", psr_payload)
print("deflated_sharpe", dsr_payload)

rng = np.random.default_rng(0)
synthetic_returns = rng.normal(scale=0.01, size=(32, 3))
synthetic_returns[:, 0] += 0.002
pbo_payload = probability_of_backtest_overfitting(synthetic_returns, slices=4)
print("synthetic_pbo", {key: pbo_payload[key] for key in ["available", "value", "reason", "path_type"] if key in pbo_payload})
````

## Older hierarchical stack, ARIMA/SARIMAX, and GLFT

This section is the pre-v2 comparison path. It stays smoke-sized even if `SMOKE` is turned off for the v2 cells.

`build_lighter_dataset` builds `next_hour_return` sequences from `configs/staged_trial_smoke.yml`. `run_stack_forward_backward` checks `PriceLSTM`, `MetaMLP`, and `ConfidenceGRU`. `run_arima_sarimax_smoke` fits the CPU baselines on at most 80 target points. `hierarchical_training_pipeline` trains the three levels for one epoch. Quotes then go through `GLFTQuoteCalculator`, `InventoryBook`, and `GLFTBacktester` with the same wiring as `lighter_compare`.


````python
from src.data.features_all import bars_for_duration
from src.experiments.lighter_compare import (
    build_lighter_dataset,
    run_arima_sarimax_smoke,
    run_stack_forward_backward,
)
from src.market_maker.glft import GLFTParams, GLFTQuoteCalculator
from src.market_maker.inventory import InventoryBook
from src.simulator.backtest import BacktestParams, GLFTBacktester
from src.training.trainer import hierarchical_training_pipeline

legacy_dataset = build_lighter_dataset(staged_smoke_config, "next_hour_return", smoke=True)
print("legacy_dataset", {
    "samples": int(legacy_dataset["X"].shape[0]),
    "X": tuple(legacy_dataset["X"].shape),
    "sequence_length": int(legacy_dataset["sequence_length"]),
    "input_size": int(legacy_dataset["input_size"]),
    "target": legacy_dataset["target"],
})
print("stack", run_stack_forward_backward(
    int(legacy_dataset["input_size"]),
    int(legacy_dataset["sequence_length"]),
    device,
))
print("arima_sarimax", run_arima_sarimax_smoke(legacy_dataset["y_ret"].numpy(), max_points=80))

hierarchical = hierarchical_training_pipeline(
    legacy_dataset["X"],
    legacy_dataset["y_ret"],
    legacy_dataset["y_dir"],
    legacy_dataset["sample_weights"],
    input_size=int(legacy_dataset["input_size"]),
    hidden_size=8,
    num_layers=1,
    dropout=0.0,
    batch_size=16,
    num_epochs=1,
    learning_rate=0.001,
    device=device,
    allow_cpu=device.type == "cpu",
)
print("hierarchical_losses", {
    "level_0": hierarchical["level_0"]["final_loss"],
    "level_1": hierarchical["level_1"]["final_loss"],
    "level_2": hierarchical["level_2"]["final_loss"],
})

stacked_model = hierarchical["hierarchical"]["model"]
stacked_model.eval()
with torch.no_grad():
    stacked_prediction, _stacked_confidence = stacked_model(
        legacy_dataset["X"].to(device),
        return_confidence=True,
    )
predicted_returns = stacked_prediction.detach().cpu().numpy().reshape(-1)
closes = legacy_dataset["close"]
predicted_prices = pd.Series(
    closes.to_numpy(dtype=float) * np.exp(np.clip(predicted_returns, -0.05, 0.05)),
    index=closes.index,
)
one_hour_bars = bars_for_duration(str(legacy_dataset["granularity"]), hours=1)
volatility = closes.pct_change().rolling(one_hour_bars, min_periods=2).std()
volatility = volatility.fillna(float(closes.pct_change().std() or 0.001))
glft_cfg = staged_smoke_config["glft"]
glft_params = GLFTParams(
    gamma=float(glft_cfg["gamma"]),
    kappa=float(glft_cfg["kappa"]),
    sigma=float(max(volatility.mean(), 1e-6)),
    dt=float(glft_cfg["dt"]),
    max_inventory=float(glft_cfg["max_inventory"]),
    min_spread=float(glft_cfg["min_spread"]),
)
backtester = GLFTBacktester(
    BacktestParams(
        start_date=pd.to_datetime(closes.index[0]).to_pydatetime(),
        end_date=pd.to_datetime(closes.index[-1]).to_pydatetime(),
        initial_capital=float(glft_cfg["initial_capital"]),
        seed=int(glft_cfg["seed"]),
        gamma=float(glft_cfg["gamma"]),
        inventory_limit=float(glft_cfg["max_inventory"]),
        quote_spread=float(glft_cfg["min_spread"]),
    ),
    GLFTQuoteCalculator(glft_params),
    InventoryBook(max_position=float(glft_cfg["max_inventory"]), max_drawdown=float(glft_cfg["max_drawdown"])),
)
glft_result = backtester.run(
    pd.DataFrame({"price": closes}),
    predicted_prices,
    volatility=volatility,
)
print("glft_metrics", glft_result.metrics)
print("glft_trades", len(glft_result.trades))
````

## Saved runs

These cells only read manifests under `artifacts/runs`. Nothing is retrained and nothing is written. The comparison is between two completed v2 meta-label runs that used different trial budgets, so a failed reproducibility check is the expected report from `compare_run_manifests`.


````python
from src.experiments.reproducibility import compare_run_manifests

def summarize_run(path: Path) -> dict:
    payload = json.loads(path.read_text(encoding="utf-8"))
    summary = {
        "path": str(path.resolve().parent.relative_to(ROOT)),
        "run_id": payload.get("run_id"),
        "status": payload.get("status"),
        "smoke": payload.get("smoke"),
        "device": payload.get("device"),
        "target": payload.get("target"),
    }
    stage1 = payload.get("stage1") or {}
    stage2 = payload.get("stage2") or {}
    if stage1:
        summary["stage1_best_trial"] = stage1.get("best_trial_id")
        summary["stage1_selection"] = stage1.get("selection")
    if stage2:
        summary["stage2_best_strategy"] = stage2.get("best_strategy_id")
        summary["stage2_selection"] = stage2.get("selection")
    best = payload.get("best_trial") or {}
    if best:
        validation = (best.get("metrics") or {}).get("validation") or {}
        summary["best_trial_id"] = best.get("trial_id")
        summary["best_trial_status"] = best.get("status")
        summary["validation_trades"] = validation.get("trades")
        summary["validation_net_pnl"] = validation.get("net_pnl")
    return summary

staged_manifest = Path("artifacts/runs/staged_5m_smoke_20260923T194754Z/manifest.json")
short_manifest = Path("artifacts/runs/meta_label_mvp_gpu_short_20260629T181752Z/manifest.json")
grid_manifest = Path("artifacts/runs/meta_label_mvp_gpu_12trial_backtest_20260629T183112Z/manifest.json")
print("staged_smoke", summarize_run(staged_manifest))
print("meta_short", summarize_run(short_manifest))
print("meta_grid", summarize_run(grid_manifest))

comparison = compare_run_manifests(short_manifest, grid_manifest)
print("compare_status", comparison["status"])
print("compare_checks", comparison["checks"])
print("compare_split_diff_trials", sorted(comparison["split_diffs"]))
print("compare_metric_diff_trials", sorted(comparison["metric_diffs"]))
````
